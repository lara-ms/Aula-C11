### **Questão 1**

Por meio do dataset paises.csv, trace dois gráficos de linhas em
um mesmo plano cartesiano, um mostrando a taxa de mortalidade
(Deathrate) e outro a taxa de natalidade (Birthrate) dos países da

### **Questão 2**

Por meio do dataset space.csv, trace um gráfico em barras
mostrando quantas empresas espaciais diferentes os EUA e a CHINA
possuem;
Dica: não se esqueça de retirar os resultados repetidos

```python
d_space = pd.read_csv('space.csv', delimiter=';')
display(d_space)

print("-"*100)
print(d_space.columns)
```

```python
emp_eua = d_space[d_space['Location'].str.contains('USA', na=False)]
emp_china = d_space.loc[d_space['Location'].str.contains('China', na=False)]

qnt_eua = emp_eua['Company Name'].nunique()
# print(qnt_eua)
qnt_china = emp_china['Company Name'].nunique()
# print(qnt_china)

plt.bar(['EUA', 'China'], [qnt_eua, qnt_china])
plt.show()
```

### **Questão 3**

Por meio do dataset space.csv, trace um gráfico em torta
ilustrando a porcentagem de missões da empresa Roscosmos que
deram certo e que deram errado;

```python
miss_roscosmos = d_space[d_space['Company Name'].str.contains('Roscosmos', na=False)]
# display(miss_roscosmos)

qnt_mr = len(miss_roscosmos[miss_roscosmos['Status Mission'] == 'Success'])
# print(qnt_mr)
qnt_mf = len(miss_roscosmos[miss_roscosmos['Status Mission'] == 'Failure'])
# print(qnt_mf)

plt.pie(x=[qnt_mr, qnt_mf],
        labels=['% Missões com sucesso',
                '% Missões com falha'],
        autopct='%1.2f%%')

plt.show()
```

### **Questão 4**

Utilizando o dataset space.csv, trace um gráfico em barras com as

### Questão 1

 Utilizando o dataset Iris do Seaborn, para os dados apenas da espécie ‘setosa’, avalie a correlação entre suas variáveis numéricas, apresentando
os valores das correlações e permitindo identificar quais variáveis estão
mais correlacionadas;

```python
ds_iris = sns.load_dataset('iris')
# display(ds_iris)
```

```python
# pegando apenas a especie setosa
setosa = ds_iris[ds_iris['species'] == 'setosa']

# selecionanod as colunas para identificar a correlacao
corr = ds_iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']].corr()

sns.heatmap(
    corr,
    annot = True,
    fmt = '.2f',
    cmap = 'Reds' # pesquisei que da para mudar a cor do heatmap
    )

plt.show()
```

### Questão 6

Com o dataset paises.csv, trace um Boxplot comparando a distribuição do “GDP ($ per capita)” entre os países da América Latina e da Europa
Ocidental, permitindo assim identificar mediana, dispersão e outliers de
cada grupo;

```python
# print(ds_paises['Region'].unique())

latina = ds_paises[ds_paises['Region'].str.strip() == 'LATIN AMER. & CARIB']
# display(latina)

europa = ds_paises[ds_paises['Region'].str.strip() == 'WESTERN EUROPE']
# display(europa)

latina_europa = pd.concat([latina, europa])
# display(latina_europa)

sns.boxplot(
    data = ds_paises,
    x = latina_europa['Region'],
    y = latina_europa['GDP ($ per capita)'],
    hue = latina_europa['Region']
    )

plt.show()
```
