```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## EXERCÍCIOS (PARTE 1)

```python
d_paises = pd.read_csv('paises.csv', delimiter=';')

display(d_paises)

print("-"*100)
print(d_paises.columns)
```

### **Questão 1**

Por meio do dataset paises.csv, trace dois gráficos de linhas em
um mesmo plano cartesiano, um mostrando a taxa de mortalidade
(Deathrate) e outro a taxa de natalidade (Birthrate) dos países da
América do Norte;

```python
# .str.strip() é para tirar os espacos em brancos que estava tendo, ele tava atrapalhando a leitura dos dados
paises_an = d_paises.loc[d_paises['Region'].str.strip() == 'NORTHERN AMERICA']
# display(paises_an)

# Mostrando os dados no mesmo plano cartesiano (eixo x -> deixado para o matplot decidir)
plt.plot(paises_an['Deathrate'], '.-r', paises_an['Birthrate'], '.-b', linewidth='2', markersize=10)
plt.show()
```

### **Questão 2**

Por meio do dataset space.csv, trace um gráfico em barras
mostrando quantas empresas espaciais diferentes os EUA e a CHINA
possuem;
Dica: não se esqueça de retirar os resultados repetidos

```python
d_space = pd.read_csv('space.csv', delimiter=';')
display(d_space)

print("-"*100)
print(d_space.columns)
```

```python
emp_eua = d_space[d_space['Location'].str.contains('USA', na=False)]
emp_china = d_space.loc[d_space['Location'].str.contains('China', na=False)]

qnt_eua = emp_eua['Company Name'].nunique()
# print(qnt_eua)
qnt_china = emp_china['Company Name'].nunique()
# print(qnt_china)

plt.bar(['EUA', 'China'], [qnt_eua, qnt_china])
plt.show()
```

### **Questão 3**

Por meio do dataset space.csv, trace um gráfico em torta
ilustrando a porcentagem de missões da empresa Roscosmos que
deram certo e que deram errado;

```python
miss_roscosmos = d_space[d_space['Company Name'].str.contains('Roscosmos', na=False)]
# display(miss_roscosmos)

qnt_mr = len(miss_roscosmos[miss_roscosmos['Status Mission'] == 'Success'])
# print(qnt_mr)
qnt_mf = len(miss_roscosmos[miss_roscosmos['Status Mission'] == 'Failure'])
# print(qnt_mf)

plt.pie(x=[qnt_mr, qnt_mf],
        labels=['% Missões com sucesso',
                '% Missões com falha'],
        autopct='%1.2f%%')

plt.show()
```

### **Questão 4**

Utilizando o dataset space.csv, trace um gráfico em barras com as
5 empresas que mais tiveram missões com status “Failure”;

```python
falhas = d_space[d_space['Status Mission'] == 'Failure']
# display(mais_falhas)

mais_falhas = falhas['Company Name'].value_counts().head(5)
# display(mais_falhas)

plt.bar(mais_falhas.index, mais_falhas)
plt.show()
```

### **Questão 5**

Por meio do dataset paises.csv, trace um gráfico de dispersão
relacionando a renda per capita (GDP ($ per capita)) com a taxa de
alfabetização (Literacy (%)) dos países da América Latina. Além disso,
faça com que o tamanho de cada ponto represente a população do país;

```python
latina = d_paises[d_paises['Region'].str.strip() == 'LATIN AMER. & CARIB']
# display(latina)

plt.scatter(latina['GDP ($ per capita)'], latina['Literacy (%)'], s=latina['Population']/100000, alpha=0.6)
plt.show()
```

### **Questão 7**

Por meio do dataset paises.csv, trace em uma mesmo plano cartesiano
duas linhas com estilos customizados (marcador, cor e tipo de linhas
diferentes) comparando o “GDP ($ per capita)” e o número de “Phones

### **Questão 8**

Por meio do dataset space.csv, usando o conceito de subplot(), trace
dois gráficos de barras separados lado a lado: à esquerda, as 5 empresas
com mais missões de sucesso; à direita, as 5 empresas com mais missões
de falha.

```python
mais_sucesso = d_space[d_space['Status Mission'] == 'Success']
# display(mais_sucesso)

mais_sucesso = mais_sucesso['Company Name'].value_counts().head(5)
# display(mais_sucesso)

plt.subplot(1, 2, 1)
plt.bar(mais_sucesso.index, mais_sucesso.values)

mais_falhas = d_space[d_space['Status Mission'] == 'Failure']
# display(mais_falhas)

mais_falhas = mais_falhas['Company Name'].value_counts().head(5)
# display(mais_falhas)

plt.subplot(1, 2, 2)
plt.bar(mais_falhas.index, mais_falhas.values)
plt.show()
```
