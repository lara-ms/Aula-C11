```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Matplotlib

## Gráficos de linhas (Plot)

Gráficos de linhas -> Plot

**Para variações ao longo do tempo**

```python
# eixo x
x = np.array([0, 1, 2, 3, 4])

# eixo y
y = x*2

# legendas nos eixos x e y
plt.xlabel("Valores x")
plt.ylabel("Valores y")

# titulo do gráfico
plt.title("Gráfico de linhas")

# plotando o grafico
plt.plot(x, y)
```

### Formatando o Gráfico

Passando o format(fmt) no plot

```python
plt.plot(x, y, '*:r')
```

```python
plt.plot(x, y, '*:r', linewidth='3', markersize=20)
```

### Plotandos dois Gráficos no mesmo Plano Cartesiano

```python
y2 = x**2

plt.plot(x, y, '*:r', x, y2, 's--b', linewidth='2', markersize=10)
plt.show()
```

### Usando conceito de SubPlots (Dashboard)

Quando se cria um subplot, deve se passar quantas linhas e quantas colunas tem a grade subplot. Alem de falar qual a posicao do grafico na grade

```python

plt.subplot(1, 2, 1)
plt.plot(x, y, '*:r', linewidth='2', markersize=10)

plt.subplot(1, 2, 2)
plt.plot(x, y2, 's--b', linewidth='2', markersize=10)
plt.show()
```

## Gráfico de Dispeção (Scatter Plot)

**Para saber a dispeção(distancia) dos dados**

```python
ds = pd.read_csv('paises.csv', delimiter=';')

ds.head()
```

```python
print(ds.columns)
```

```python
# Quao dispersas estao as rendas per capita dos 6 maiores paises do mundo?

# Pegando apenas os 6 maiores paises do planeta
maiores = ds.nlargest(6, 'Area (sq. mi.)')

maiores
```

```python
# Tracando o grafico com as rendas per capitas desses paises
plt.scatter(maiores['Country'], maiores['GDP ($ per capita)'])
plt.show()
```

```python
# Adicionando uma terceira dimensao
# O parametro S permite adicionar uma terceira dimensao neste grafico
plt.scatter(maiores['Country'],
            maiores['GDP ($ per capita)'],
            s=maiores['Area (sq. mi.)']/10000)
plt.show()
```

## Gráfico em Barras (Bar Plot)

**Compara a quantidade de dados em cada coisa (dados quantidados)**

```python
# Qual a diferenca das 5 maiores rendas per capita do dataset?

maiores = ds.nlargest(5, 'GDP ($ per capita)')
maiores
```

```python
plt.bar(maiores['Country'], maiores['GDP ($ per capita)'])
plt.show()
```

## Gráfico em Torta/Pizza (Pie Plot)

**Mostra as proporções dos dados**

```python
# Pegando os paises que nao possuem costa maritma
paises_semCosta = ds[ds['Coastline (coast/area ratio)'] == 0]
paises_semCosta
```

```python
# Calculando a quantidade e paises sem costa e quantidade com costa
quant_semCosta = len(paises_semCosta)
quant_comCosta = len(ds) - quant_semCosta

print(quant_semCosta, quant_comCosta)
```

```python
plt.pie(x=[quant_semCosta, quant_comCosta],
        labels=['% Paises sem costa',
                '% Paises com costa'],
        autopct='%1.2f%%')

plt.show()
```

## Fundamentos de Seaborn

```python
# importando o dataset tips -> disponivel no seaborn
ds_tips = sns.load_dataset('tips')

display(ds_tips)
```

## Scatter Plot

```python
sns.scatterplot(data = ds_tips, x = 'total_bill', y = 'tip')

# Para dar detalhes usa o matplotlib
plt.xlabel('Total da conta em US$')
plt.ylabel('Gorjeta em US$')
plt.show()
```

```python
sns.scatterplot(data = ds_tips, x = 'total_bill', y = 'tip')

# Setando um estilo diferente no grafico
sns.set_style('dark')

# Setando uma fonte maior
sns.set_context('talk')

# Para dar detalhes usa o matplotlib
plt.xlabel('Total da conta em US$')
plt.ylabel('Gorjeta em US$')
plt.show()
```

## Histogramas


> Distribuição de dados numéricos em intervalos **(bins)**



```python
ds_penguins = sns.load_dataset('penguins')

display(ds_penguins)

print('-'*100)
print(ds_penguins.columns)
```

```python
# tracando o histograma
sns.histplot(data = ds_penguins,
             x = 'flipper_length_mm',
             hue = 'species', # (hue -> dado categórico)
             kde = True # um plot sobre os dados, distribuicao de probabilidade dos dados
             )

plt.show()
```

## BoxPlot



> Ver a **mediana (linha dentro da caixa)** de um conjunto de dados

> Mostra **outliers (pontos fora da curva)**

> **50% dos dados** estão dentro da caixa



```python
ds_tips = sns.load_dataset('tips')

display(ds_tips)
```

```python
# tracando o boxplot
sns.boxplot(
    data = ds_tips,
    x = 'day',
    y = 'tip',
    hue = 'sex' # dado categorico
    )
plt.show()
```

## Regplot

> Mostra a **tendencia** dos dados por meio de uma **regressão**

```python
ds_tips = sns.load_dataset('tips')

display(ds_tips)
```

```python
# tracando o regplot
sns.regplot(
    data = ds_tips,
    x = 'total_bill',
    y = 'tip',
    line_kws = {'color': 'purple'} # definindo a cor da linha de regressão
    )
plt.show()
```

## Heatmap

> **Correlação** mede exatamente a **relação** entre duas coisas de **forma quantificada**
- **Próximo de 1** : correlação **mais forte**
- **Próximo de 0** : correlação **mais fraca**

```python
# selecionanod as colunas que quero identificar uma correlacao
corr = ds_tips[['total_bill', 'tip', 'size']].corr()

# tracando o heatmap
sns.heatmap(
    corr,
    annot=True, # mostra os valores dentro das celulas
    fmt='.2f', # formatando os valores para duas casas decimais
    cmap='Purples' # definindo a cor (tem q colocar no plural)
)
plt.show()
```
